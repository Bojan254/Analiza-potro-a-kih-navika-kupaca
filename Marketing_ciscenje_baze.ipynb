```python
import pandas as pd
import numpy as np

df = pd.read_csv("marketing_campaign.csv", sep="\t")
```

```python
print(df.shape)           # koliko redova i kolona ima
print(df.isnull().sum())  # broj null vrijednosti po koloni

# Popuni null vrijednosti u Income prosjekom
df["Income"].fillna(df["Income"].mean(), inplace=True)
```

```python
df["Age"] = 2015 - df["Year_Birth"]
```

```python
bins_age = [0, 30, 40, 50, 60, 999]
labels_age = [
    "Do 30",
    "Od 30 do 40",
    "Od 40 do 50",
    "Od 50 do 60",
    "Preko 60"
]
df["Year_category"] = pd.cut(
    df["Age"],
    bins=bins_age,
    labels=labels_age,
    right=False
)
## razvrstavanje po starosnim grupama
```

```python
bins_inc = [0, 30000, 50000, 70000, 100000, 9999999]
labels_inc = [
    "Do 30k",
    "Od 30k do 50k",
    "Od 50k do 70k",
    "Od 70k do 100k",
    "Preko 100k"
]
df["Income_category"] = pd.cut(
    df["Income"],
    bins=bins_inc,
    labels=labels_inc,
    right=False
)
```

```python
df["Total_spent"] = (
    df["MntWines"] +
    df["MntFruits"] +
    df["MntMeatProducts"] +
    df["MntFishProducts"] +
    df["MntSweetProducts"] +
    df["MntGoldProds"]
)
```

```python
df["Total_purchases"] = (
    df["NumWebPurchases"] +
    df["NumCatalogPurchases"] +
    df["NumStorePurchases"]
)
```

```python
conditions = [
    df["NumDealsPurchases"] == 0,
    df["NumDealsPurchases"].between(1, 2),
    df["NumDealsPurchases"].between(3, 4),
    df["NumDealsPurchases"].between(4, 6),
    df["NumDealsPurchases"] > 6
]
choices = [
    "Nikad na popustu",
    "1-2 puta",
    "3-4 puta",
    "4-6 puta",
    "Preko 6 puta"
]
df["Deals_category"] = np.select(
    conditions, choices, default="Nepoznato"
)
```
